```python
def calc_sales_tax(price, tax_rate=0.13):
    tax_amt = price * tax_rate
    return tax_amt
```

```python
calc_sales_tax(6,5)
```

```python
def calc_total_bill(price, tax_rate=0.13, tip_rate=0.2):
    tax_amt = price * tax_rate
    tip_amt = price * tip_rate
    total_cost = price + tax_amt + tip_amt
    return total_cost
```

```python
calc_total_bill(100)
```

```python
calc_total_bill(100,0.15)
```

```python
calc_total_bill(100, tip_rate=0.25)
```

```python
# break statement to interrupt a loop
countdown = 4
while countdown >0:
    print(countdown)
    if countdown ==3:
        print('we are breaking out of the loop early!')
        break
    countdown -=1

print('Done iterating')
```
